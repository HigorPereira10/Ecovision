```python
import os
import glob
import random
import shutil

# Monta o Drive quando formos rodar no Colab
from google.colab import drive
drive.mount('/content/drive')

# Definimos a raiz de onde extraímos os dados no Passo 2
BASE_PATH = "/content/drive/MyDrive/Colab Notebooks/TomatoMAP/Dataset_Fenologia"
```

```python
# Busca todas as pastas que começam com "plant" ignorando as subpastas de datas
caminho_busca = os.path.join(BASE_PATH, "img", "*", "*", "plant*")
pastas_plantas = glob.glob(caminho_busca)

# Extrai apenas o nome final (ex: 'plant1', 'plant10') e remove duplicatas
plantas_unicas = list(set([os.path.basename(p) for p in pastas_plantas]))
plantas_unicas.sort() # Organiza em ordem alfabética/numérica

print(f"Total de plantas físicas identificadas: {len(plantas_unicas)}")
print(f"Exemplos de IDs: {plantas_unicas[:5]}")
```

```python
# Fixamos a 'seed' para garantir que o sorteio seja sempre o mesmo se precisarmos rodar de novo
random.seed(42)

# Embaralha a lista de plantas
random.shuffle(plantas_unicas)

total = len(plantas_unicas)
idx_treino = int(total * 0.70)
idx_val = int(total * 0.85)

# Fatiamento das listas
plantas_treino = plantas_unicas[:idx_treino]
plantas_validacao = plantas_unicas[idx_treino:idx_val]
plantas_teste = plantas_unicas[idx_val:]

print("=== DISTRIBUIÇÃO DO SPLIT POR PLANTA ===")
print(f"Treino (70%): {len(plantas_treino)} plantas")
print(f"Validação (15%): {len(plantas_validacao)} plantas")
print(f"Teste (15%): {len(plantas_teste)} plantas")

# Exemplo de verificação para garantir que não há vazamento de dados
intersecao = set(plantas_treino).intersection(set(plantas_teste))
print(f"\nVazamento de dados (Plantas no treino e no teste simultaneamente): {len(intersecao)}")
```
